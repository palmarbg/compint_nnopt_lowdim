# Download datasets

```python
import kagglehub
import os.path
import shutil

rpath_datafolder = './data'
```

```python
def download_dataset(kaggle_token, rpath_savepath):
    tmp_path = kagglehub.dataset_download(kaggle_token)

    shutil.move(tmp_path, rpath_savepath)
```

## Fashion MNIST

```python
download_dataset("zalando-research/fashionmnist", os.path.join(rpath_datafolder, 'fashionmnist'))
```

Delete unnecessary files.

```python
%%bash
cd data/fashionmnist/
for f in *; do
  [[ $f == *.csv ]] || rm -f "$f"
done
```
